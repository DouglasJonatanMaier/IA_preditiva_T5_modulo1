```python
# ingressar um numero > 1 e imprimir seus divisores

numero =  int(input("Ingressar numero"))

divisor = 1
print("NUmero:", numero)
while divisor < numero:
    if numero % divisor == 0:
        print(divisor, end=" ")
    divisor += 1
```

```python
# ingressar um numero > 1 e imprimir seus divisores

numero =  int(input("Ingressar numero"))

divisor = 1
contadr_divisores = 0
print("NUmero:", numero)
while divisor < numero:
    if numero % divisor == 0:
        contadr_divisores += 1
    divisor += 1
print("O numero de divisores é", contadr_divisores)
```
