```python
"""Exercício 2: Análise de vendas da semana
Problema
Uma pequena loja quer analisar suas vendas durante uma semana.
Para cada dia, o vendedor informar o valor total vendido naquele dia.
Você deve calcular:
O total vendido durante a semana;
A média de vendas por dia;
Qual foi o dia com maior venda;
Quantos dias tiveram vendas acima da média.

Considere:
Segunda-feira = dia 1
Terça-feira = dia 2
...
Domingo = dia 7

Entrada
Serão informados 7 valores, um para cada dia da semana.

Saída
Imprima:
o total vendido;
a média;
o dia de maior venda;
a quantidade de dias acima da média."""

# entrada = 100 200 150 300 250 180 220
# Total: 1400 
# Media: 200.0 
# Melhor dia: 4 
# Dias acima da media: 3 

total_sem = 0
media_aritmetica = 0 # soma media
id_maior = 1 
valor_maior = 0

valor_sem = [100, 200, 150, 300, 250, 180, 220 ]

for i in range(len(valor_sem)):
    total_sem += valor_sem[i]
    if valor_maior < valor_sem[i]:
        valor_maior = valor_sem[i]
        id_maior = i

media_aritmetica = total_sem / len(valor_sem)

cont = 0

for j in range(len(valor_sem)):
    if valor_sem[j] >media_aritmetica:
        cont += 1


print("Total da semana é", total_sem)
print("Media aritmetica é", media_aritmetica)
print("Melhor dia foi o dia", id_maior+1)
print("Dias acima da media foram", cont)


```

```python
nomes = ["carla", "Jose", "Artur", "Rosally"]
idades = [24, 37, 33, 18]
list_nome_coluna = ["Nomes", "Idades"]
list_index = ["Pessoa 1", "Pessoa 2", "Pessoa 3", "Pessoa 4"]

dados = list(zip(nomes, idades))
print(dados)
print("")

df_dados = pd.DataFrame(data= dados, columns= list_nome_coluna, index= list_index)

print(df_dados)
```
