```python
"""Problema
Uma família deseja analisar o consumo de energia elétrica durante 7 dias.
Para cada dia, é informado o consumo de energia em kWh.
O programa deve informar:
O consumo total da semana;
O consumo médio diário;
O dia de maior consumo;
Quantos dias tiveram consumo acima da média;
Quantos dias tiveram consumo menor ou igual a 10 kWh.

Entrada
Serão informados 7 valores, um para cada dia.
Dia 1 → segunda-feira
Dia 2 → terça-feira
...
Dia 7 → domingo

Saída
Imprima os 5 items pedidos na descripção"""

# entrada dias :8 12 15 9 20 11 7

consumo = [8, 12, 15, 9, 20, 11, 7]
kwh_semana = 0
kwh_medio = 0
cont = 0
dia_maior = 0
kwh_maior = 0

    
while cont < len(consumo):
    kwh_semana += consumo[cont]

    if kwh_maior < consumo[cont]:
        kwh_maior = consumo[cont]
        dia_maior += cont

    cont += 1

kwh_medio = kwh_semana/ len(consumo)

print (kwh_semana)
print(kwh_medio)
print(dia_maior)

```

```python
# argumentos
def imprimir_mensagem(lista_local):
    lista_local[-1] = 10
    print(lista_local)
lista_a = [1, 2, 3, 4, 5]
print(lista_a)
imprimir_mensagem(lista_a.copy())
print(lista_a)
```

```python
# enviar estrutura de dados na função
def func_soma_lista(var_lista):
    soma = 0
    for i in range(len(var_lista)):
        soma += var_lista[i]

    return soma

soma_a = func_soma_lista([1, 3, 23, 344,1, 4])
print("Soma =", soma_a)
```
