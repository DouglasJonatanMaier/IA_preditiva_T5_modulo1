```python
#B elevado E = B x B x B ... B
B = float(input())
E = int(input())

contador = 1
valor_final = 1

while contador <= E:
    valor_final = valor_final * B # valor final * B
    contador += 1 #contador + 1

print("valor final", valor_final)
```

```python
#faça um laço de repetição que printe os numeros de 0 a 100 porém pulando de 2 em 2

# range (inicio, final, passo)
for valor in range(0, 101, 2):
    print(valor, end = " ")

print("") #adicionar salto de linha.
contador = 0 # inicio
while contador < 101: #final vai ser a condição.
    print(contador, end=" ")
    contador += 2 # passo
```

```python
#faça um laço de repetição que irá rodar 5 vezes
#a cada volta ele terá que usar um input para pedir um numero
#mas somente printar em tela os numeros maiores que 10

for i in range (5): # [0, 1, 2, 3, 4,] rodar 5 vezes
    numero = float(input("ingresse um numero: "))# pedir numero
    if numero > 10: # maior que 10
        print(numero)

print("Final do programa")

contador = 0
while contador <5: # repetir (inicio)
    numero_while = float(input("ingressar numero: ", )) #pedir numero
    if numero_while > 10: # final
        print(numero_while)
    contador += 1# passo
print("Final do While")
```
