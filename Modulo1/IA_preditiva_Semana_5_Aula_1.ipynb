```python
"""Exercício 1: Controle de fila no mercado 

Problema
Um supermercado possui uma fila de clientes. Para cada cliente, é informado o número de produtos que ele possui no carrinho.
O caixa consegue atender um produto por segundo.
Você deve determinar:
Quanto tempo será necessário para atender todos os clientes;
Qual cliente possui o maior número de produtos;
Quantos clientes possuem 10 produtos ou mais.

Entrada
4 12 7 15 3
Uma linha contém números inteiros, representando a quantidade de produtos de cada cliente.

Saída
Imprima:
o tempo total de atendimento em segundos;
o número do cliente com mais produtos;
a quantidade de clientes com 10 ou mais produtos.

Considere que o primeiro cliente é o cliente 1."""


lista_cliente = [4, 12, 7, 15, 3]

tempo_total = 0
id_cliente_mais_produto = 0
max_qnt_prod = lista_cliente[0]
cliente_mais_10_produtos = 0

for id in range(len(lista_cliente)): #[0,1,2,3]
    #contagem de tempo total de caixa.
    tempo_total = tempo_total + lista_cliente[id]
    # cliente com mais de 10 produtos.
    if max_qnt_prod < lista_cliente[id]:
        max_qnt_prod = lista_cliente[id]
        id_cliente_mais_produto = id
    # clientes com mais de 10 produtos.
    if lista_cliente[id] >= 10:
        cliente_mais_10_produtos += 1
        

print("tempo total é :", tempo_total)
print("Cliente com mais produtos :", id_cliente_mais_produto)
print("Clientes com 10 ou mais produtos: ", cliente_mais_10_produtos)
```

```python
lista_cliente = [2, 3, 4, 5]

tempo_total = 0
id_cliente_mais_produto = 0
max_qnt_prod = lista_cliente[0]
cliente_mais_10_produtos = 0
id = 0

while id < len(lista_cliente):
    # contagem de tempo total de caixa.
    tempo_total = tempo_total + lista_cliente[id]
    # cliente com mais de 10 produtos.
    if max_qnt_prod < lista_cliente[id]:
        max_qnt_prod = lista_cliente[id]
        id_cliente_mais_produto = id
    # clientes com mais de 10 produtos.
    if lista_cliente[id] >= 10:
        cliente_mais_10_produtos += 1
    id += 1

print("tempo total é :", tempo_total)
print("Cliente com mais produtos :", id_cliente_mais_produto)
print("Clientes com 10 ou mais produtos: ", cliente_mais_10_produtos)
```
